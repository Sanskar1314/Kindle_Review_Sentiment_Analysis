## About Dataset
Context This is a small subset of dataset of Book reviews from Amazon Kindle Store category.

Content 5-core dataset of product reviews from Amazon Kindle Store category from May 1996 - July 2014. Contains total of 982619 entries. Each reviewer has at least 5 reviews and each product has at least 5 reviews in this dataset. Columns

- asin - ID of the product, like B000FA64PK
- helpful - helpfulness rating of the review - example: 2/3.
- overall - rating of the product.
- reviewText - text of the review (heading).
- reviewTime - time of the review (raw).
- reviewerID - ID of the reviewer, like A3SPTOKDG7WBLN
- reviewerName - name of the reviewer.
- summary - summary of the review (description).
- unixReviewTime - unix timestamp.

Acknowledgements This dataset is taken from Amazon product data, Julian McAuley, UCSD website. http://jmcauley.ucsd.edu/data/amazon/

License to the data files belong to them.

```python
df['reviewText']=df['reviewText'].str.lower()
```

```python
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
from bs4 import BeautifulSoup
```

```python
## Removing special characters
df['reviewText']=df['reviewText'].apply(lambda x:re.sub('[^a-z A-z 0-9-]+', '',x))
## Remove the stopswords
df['reviewText']=df['reviewText'].apply(lambda x:" ".join([y for y in x.split() if y not in stopwords.words('english')]))
## Remove url 
df['reviewText']=df['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))
## Remove html tags
df['reviewText']=df['reviewText'].apply(lambda x: BeautifulSoup(x, 'lxml').get_text())
## Remove any additional spaces
df['reviewText']=df['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
df.head()
```

```python
# Lemmatizer 
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
def lemmatize_words(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
df['reviewText']=df['reviewText'].apply(lambda x:lemmatize_words(x))
```

```python
df.head()
```

```python
# TrainTest Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df['reviewText'],df['rating'],test_size=0.2)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow=CountVectorizer()
X_train_bow=bow.fit_transform(X_train).toarray()
X_test_bow=bow.transform(X_test).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
X_train_tfidf=tfidf.fit_transform(X_train).toarray()
X_test_tfidf=tfidf.transform(X_test).toarray()
```

```python
X_train_bow
```

```python
X_train_tfidf
```

```python
from sklearn.naive_bayes import GaussianNB
nb_model_bow=GaussianNB().fit(X_train_bow,y_train)
nb_model_tfidf=GaussianNB().fit(X_train_tfidf,y_train)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
y_pred_bow=nb_model_bow.predict(X_test_bow)
```

```python
y_pred_tfidf=nb_model_tfidf.predict(X_test_tfidf)
```

```python
print("Bow Accuracy:",accuracy_score(y_test,y_pred_bow))
```

```python
print("Tfidf Accuracy:",accuracy_score(y_test,y_pred_tfidf))
```

## Word2Vec & Average Word2Vec
Train a Word2Vec model on the review corpus, then represent each review as the **average** of its word vectors.

```python
from gensim.models import Word2Vec
import numpy as np
```

```python
# Tokenize each review into a list of words
X_train_tokens = [text.split() for text in X_train]
X_test_tokens = [text.split() for text in X_test]

print(f'Training samples: {len(X_train_tokens)}')
print(f'Test samples: {len(X_test_tokens)}')
print(f'Example tokens: {X_train_tokens[0][:10]}')
```

```python
# Train Word2Vec model on the training data
w2v_model = Word2Vec(
    sentences=X_train_tokens,
    vector_size=100,   # Dimensionality of word vectors
    window=5,          # Context window size
    min_count=2,       # Ignore words appearing less than 2 times
    workers=4,         # Parallel threads
    epochs=20          # Training iterations
)

print(f'Vocabulary size: {len(w2v_model.wv.key_to_index)}')
print(f'Vector size: {w2v_model.wv.vector_size}')
```

```python
# Quick sanity check - most similar words
w2v_model.wv.most_similar('book')
```

```python
def avg_word2vec(tokenized_texts, model, vector_size):
    """
    For each document (list of tokens), compute the average of the
    word vectors for all tokens that exist in the Word2Vec vocabulary.
    If none of the tokens are in the vocabulary, return a zero vector.
    """
    vocab = model.wv.key_to_index
    features = []
    for tokens in tokenized_texts:
        word_vectors = [model.wv[word] for word in tokens if word in vocab]
        if len(word_vectors) > 0:
            features.append(np.mean(word_vectors, axis=0))
        else:
            features.append(np.zeros(vector_size))
    return np.array(features)
```

```python
# Build Average Word2Vec feature matrices
X_train_w2v = avg_word2vec(X_train_tokens, w2v_model, 100)
X_test_w2v = avg_word2vec(X_test_tokens, w2v_model, 100)

print(f'X_train_w2v shape: {X_train_w2v.shape}')
print(f'X_test_w2v shape: {X_test_w2v.shape}')
```

```python
# Train Gaussian Naive Bayes on Average Word2Vec features
nb_model_w2v = GaussianNB().fit(X_train_w2v, y_train)
```

```python
y_pred_w2v = nb_model_w2v.predict(X_test_w2v)
```

```python
print('Word2Vec Accuracy:', accuracy_score(y_test, y_pred_w2v))
print('\nClassification Report (Word2Vec):')
print(classification_report(y_test, y_pred_w2v))
```

```python
# Compare all approaches
print('='*50)
print('Accuracy Comparison')
print('='*50)
print(f'BOW Accuracy:      {accuracy_score(y_test, y_pred_bow):.4f}')
print(f'TF-IDF Accuracy:   {accuracy_score(y_test, y_pred_tfidf):.4f}')
print(f'Word2Vec Accuracy: {accuracy_score(y_test, y_pred_w2v):.4f}')
```
